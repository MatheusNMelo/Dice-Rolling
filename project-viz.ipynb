#Imports

```python
import plotly.express as px
import plotly.graph_objects as go
import plotly.subplots as sp
import pandas as pd
import numpy as np
```

```python
df1_results = pd.read_csv("df1_results.csv",index_col=0)
df2_results = pd.read_csv("df2_results.csv",index_col=0)
df1_results.drop("Size",inplace=True)
df2_results.drop("Size",inplace=True)
```

```python
df1_results
```

```python
df2_results
```


# 1. Gráfico de Barras para Valores-p Médios

```python

fig1 = go.Figure()
fig1.add_trace(go.Bar(
    x=df1_results.index,
    y=df1_results["Average"],
    name="NIST Beacon",
    marker_color='blue'
))
fig1.add_trace(go.Bar(
    x=df2_results.index,
    y=df2_results["Average"],
    name="Local Beacon",
    marker_color='red'
))
fig1.update_layout(
    title="Valores-p Médios por Teste",
    xaxis_title="Testes",
    yaxis_title="Valor-p Médio",
    barmode='group'
)
fig1.show()

```


# 2. Gráfico de Linhas para Valores-p por Parte


```python

fig2 = go.Figure()
for i in range(4):
    fig2.add_trace(go.Scatter(
        x=df1_results.index,
        y=df1_results[f"Part {i+1}"],
        name=f"NIST Beacon - Part {i+1}",
        mode='lines+markers'
    ))
    fig2.add_trace(go.Scatter(
        x=df2_results.index,
        y=df2_results[f"Part {i+1}"],
        name=f"Local Beacon - Part {i+1}",
        mode='lines+markers'
    ))
fig2.update_layout(
    title="Valores-p por Parte",
    xaxis_title="Testes",
    yaxis_title="Valor-p",
    legend_title="Partes"
)
fig2.show()
```


# 3. Heatmap de Valores-p

```python

heatmap_data = df1_results.iloc[:, :-2].T  # Excluir colunas de média e desvio padrão
fig3 = px.imshow(
    heatmap_data,
    labels=dict(x="Testes", y="Partes", color="Valor-p"),
    x=df1_results.index,
    y=[f"Part {i+1}" for i in range(4)],
    title="Heatmap de Valores-p (NIST Beacon)"
)
fig3.show()
```


# 4. Gráfico de Dispersão para Média vs. Desvio Padrão

```python
fig4 = go.Figure()
fig4.add_trace(go.Scatter(
    x=df1_results["Average"],
    y=df1_results["Std Dev"],
    mode='markers',
    name="NIST Beacon",
    marker=dict(color='blue', size=10)
))
fig4.add_trace(go.Scatter(
    x=df2_results["Average"],
    y=df2_results["Std Dev"],
    mode='markers',
    name="Local Beacon",
    marker=dict(color='red', size=10)
))
fig4.update_layout(
    title="Média vs. Desvio Padrão dos Valores-p",
    xaxis_title="Média dos Valores-p",
    yaxis_title="Desvio Padrão dos Valores-p",
    legend_title="Beacon"
)
fig4.show()
```

# 5. Gráfico de Caixa (Box Plot)


```python
fig5 = go.Figure()
for i in range(4):
    fig5.add_trace(go.Box(
        y=df1_results[f"Part {i+1}"],
        name=f"NIST Beacon - Part {i+1}",
        boxpoints='all',
        jitter=0.3,
        pointpos=-1.8
    ))
    fig5.add_trace(go.Box(
        y=df2_results[f"Part {i+1}"],
        name=f"Local Beacon - Part {i+1}",
        boxpoints='all',
        jitter=0.3,
        pointpos=-1.8
    ))
fig5.update_layout(
    title="Distribuição dos Valores-p por Parte",
    xaxis_title="Partes",
    yaxis_title="Valor-p",
    legend_title="Beacon"
)
fig5.show()
```
